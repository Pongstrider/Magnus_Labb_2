```python
import numpy as np
from matplotlib import pyplot as plt
import soundfile as sf
#b

ljudet, fs = sf.read("ett_a_sin.wav")


ts= 1/fs



print("Samplingsfrekvens:", fs)
print("Antal sampel:", len(ljudet))
print("Samplingstid:", ts)

```

```python
t = np.arange(len(ljudet)) / fs
```

```python
plt.plot(t, ljudet)
plt.xlabel("Tid: sekunder")
plt.ylabel("Amplitud")
plt.title("ljudsignal")
plt.show()

#blå vänster
#orange höger
```

# D)

```python
plt.plot(t, ljudet.T[0])
plt.xlim(0.01, 0.03)
plt.xlabel("Tid (sekunder)")
plt.ylabel("Amplitud")
plt.title("Inzoomad ljudsignal")
plt.grid()
plt.show()
```

```python
T = 0.02 / 4.5 # en uppskattning , det är ungefär 9 perioder på grafen ovan
L  = T * fs

print(L , "samples per period")
```

```python
kanal_v = ljudet.T[0]

nollgenomgangar = np.where(
    (kanal_v[:-1] < 0) & (kanal_v[1:] >= 0)
)[0]

print(nollgenomgangar[:25])
```
